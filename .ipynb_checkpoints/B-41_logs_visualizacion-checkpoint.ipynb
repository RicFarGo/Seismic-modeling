# Visualizacion de logs B-41

Notebook para cargar y graficar logs de pozo (`GAMMA`, `DENSITY`, `SONIC`) desde:

`C:/Users/rfaro/Documents/ParaView/B-41_logs.txt`

Parsea la seccion `~Ascii Data Section`, reemplaza nulos (`-999.25`) y grafica tracks con formato tipo petrofisico.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- Parametros editables ---
FILE_PATH = r"C:\Users\rfaro\Documents\ParaView\B-41_logs.txt"
NULL_VALUE = -999.25

# Intervalo para parecerse a la imagen de referencia
DEPTH_MIN = 0
DEPTH_MAX = 2300.0

# Limites de tracks
GAMMA_LIMS = (0, 200)
DENSITY_LIMS = (1.95, 2.95)
SONIC_LIMS = (140, 15)  # invertido horizontalmente

# Colores solicitados
GAMMA_COLOR = "green"
DENSITY_COLOR = "red"
SONIC_COLOR = "darkblue"
```

```python
def parse_ascii_las_like(file_path: str, null_value: float = -999.25) -> pd.DataFrame:
    """Parsea un archivo LAS-like y devuelve DataFrame con columnas estandar."""
    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        lines = f.readlines()

    # Buscar inicio de seccion ascii
    start_idx = None
    for i, line in enumerate(lines):
        if line.strip().upper().startswith("~ASCII"):
            start_idx = i + 1
            break

    if start_idx is None:
        raise ValueError("No se encontro la seccion ~Ascii Data Section en el archivo.")

    rows = []
    for line in lines[start_idx:]:
        text = line.strip()
        if not text or text.startswith("#") or text.startswith("~"):
            continue

        parts = text.split()
        # Esperamos 5 columnas: DEPT, DT, GR, RHOB, PVEL
        if len(parts) < 5:
            continue

        try:
            row = [float(parts[0]), float(parts[1]), float(parts[2]), float(parts[3]), float(parts[4])]
            rows.append(row)
        except ValueError:
            continue

    if not rows:
        raise ValueError("No se pudieron extraer datos numericos del bloque ASCII.")

    df = pd.DataFrame(rows, columns=["DEPT", "DT", "GR", "RHOB", "PVEL"])

    # Reemplazar nulos del archivo por NaN
    df = df.replace(null_value, np.nan)

    return df
```

```python
df = parse_ascii_las_like(FILE_PATH, NULL_VALUE)

# Curvas de trabajo
plot_df = pd.DataFrame({
    "DEPT": df["DEPT"],
    "GAMMA": df["GR"],
    "DENSITY": df["RHOB"],
    "SONIC": df["DT"],
})

# Velocidad desde sonico DT (us/ft) -> m/s
# v = 0.3048 / (DT * 1e-6) = 304800 / DT
plot_df["VEL_FROM_SONIC_MPS"] = 304800.0 / plot_df["SONIC"]

# Tiempo doble TWT (ms): 2 * profundidad / velocidad
plot_df["TWT_MS"] = (2.0 * plot_df["DEPT"] / plot_df["VEL_FROM_SONIC_MPS"]) * 1000.0

# Filtrar intervalo si se define
plot_df = plot_df[(plot_df["DEPT"] >= DEPTH_MIN) & (plot_df["DEPT"] <= DEPTH_MAX)].copy()
plot_df = plot_df.sort_values("DEPT")

print(f"Filas en ventana de profundidad: {len(plot_df):,}")
plot_df[["DEPT", "SONIC", "VEL_FROM_SONIC_MPS", "TWT_MS"]].head()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(9, 10), sharey=True)

# Track 1: GAMMA
axes[0].plot(plot_df["GAMMA"], plot_df["DEPT"], color=GAMMA_COLOR, linewidth=1.0)
axes[0].set_xlim(*GAMMA_LIMS)
axes[0].set_xlabel("GAMMA", fontsize=14, fontweight="bold")
axes[0].set_ylabel("(m)", fontsize=12)
axes[0].grid(True, alpha=0.3)

# Track 2: DENSITY
axes[1].plot(plot_df["DENSITY"], plot_df["DEPT"], color=DENSITY_COLOR, linewidth=1.0)
axes[1].set_xlim(*DENSITY_LIMS)
axes[1].set_xlabel("DENSITY", fontsize=14, fontweight="bold")
axes[1].grid(True, alpha=0.3)

# Track 3: SONIC
axes[2].plot(plot_df["SONIC"], plot_df["DEPT"], color=SONIC_COLOR, linewidth=1.0)
axes[2].set_xlim(*SONIC_LIMS)
axes[2].set_xlabel("SONIC", fontsize=14, fontweight="bold")
axes[2].grid(True, alpha=0.3)

# Profundidad hacia abajo (estandar de logs)
axes[0].invert_yaxis()

# Estilo de ticks similar al ejemplo
for ax in axes:
    ax.tick_params(axis="both", labelsize=10)

plt.tight_layout()
plt.show()
```

## Notas

- Si quieres ver toda la columna de profundidad, cambia `DEPTH_MIN` y `DEPTH_MAX`.
- Si deseas ajustar el grosor de curvas, modifica `linewidth` en cada track.
- `SONIC` usa `DT` (us/ft), como solicitaste.
- Se agrega `VEL_FROM_SONIC_MPS` calculada desde `SONIC` y `TWT_MS` (tiempo doble en milisegundos).

## Time sampling y remuestreo en TWT

Se conserva `DEPTH_SAMPLING_M` y `VEL_FROM_SONIC_MPS`, se crea `TIME_SAMPLING_SEC = DEPTH_SAMPLING_M / VEL_FROM_SONIC_MPS`, se convierte a `TWT_SAMPLING_MS`, y su media se usa como paso regular exacto (sin redondeo) para remuestrear curvas.

```python
# Base completa para remuestreo (sin recorte de ventana)
resample_df = df.sort_values("DEPT").reset_index(drop=True).copy()

# Mantener velocidad desde sonico DT (us/ft) -> m/s
resample_df["VEL_FROM_SONIC_MPS"] = 304800.0 / resample_df["DT"]

# Mantener muestreo en profundidad por punto
resample_df["DEPTH_SAMPLING_M"] = resample_df["DEPT"].shift(-1) - resample_df["DEPT"]

# Nuevo log de time sampling (one-way, segundos)
resample_df["TIME_SAMPLING_SEC"] = resample_df["DEPTH_SAMPLING_M"] / resample_df["VEL_FROM_SONIC_MPS"]

# Pasar a TWT en milisegundos (factor 2 confirmado)
resample_df["TWT_SAMPLING_MS"] = 2.0 * resample_df["TIME_SAMPLING_SEC"] * 1000.0

# TWT acumulado por muestra (util para eje temporal por traza)
valid_twt_sampling = resample_df["TWT_SAMPLING_MS"].replace([np.inf, -np.inf], np.nan)
resample_df["TWT_MS"] = valid_twt_sampling.fillna(0.0).cumsum()

# Media exacta (sin redondeo) ignorando invalidos
valid_sampling = valid_twt_sampling[np.isfinite(valid_twt_sampling) & (valid_twt_sampling > 0)]
if valid_sampling.empty:
    raise ValueError("No hay valores validos en TWT_SAMPLING_MS para calcular la media.")

mean_twt_sampling_ms = float(valid_sampling.mean())

print(f"Media de TWT_SAMPLING_MS (paso exacto): {mean_twt_sampling_ms:.6f} ms")
resample_df[["DEPT", "DEPTH_SAMPLING_M", "DT", "VEL_FROM_SONIC_MPS", "TIME_SAMPLING_SEC", "TWT_SAMPLING_MS"]].head()
```

```python
# Malla regular en TWT usando paso exacto (sin redondeo)
valid_twt = resample_df["TWT_MS"][np.isfinite(resample_df["TWT_MS"])]
twt_min = float(valid_twt.min())
twt_max = float(valid_twt.max())

dt_regular_ms = mean_twt_sampling_ms
twt_grid = np.arange(twt_min, twt_max + dt_regular_ms, dt_regular_ms)

# Remuestreo por media para todas las columnas numericas
numeric_cols = resample_df.select_dtypes(include=[np.number]).columns.tolist()
work = resample_df[np.isfinite(resample_df["TWT_MS"])].copy()
work["BIN_ID"] = np.floor((work["TWT_MS"] - twt_min) / dt_regular_ms).astype(int)

agg = work.groupby("BIN_ID", as_index=False)[numeric_cols].mean()

resampled_logs_twt = pd.DataFrame({
    "BIN_ID": np.arange(len(twt_grid), dtype=int),
    "TWT_MS_REG": twt_grid,
})
resampled_logs_twt = resampled_logs_twt.merge(agg, on="BIN_ID", how="left")

# TWT regular como columna principal
cols = ["TWT_MS_REG"] + [c for c in resampled_logs_twt.columns if c not in ["TWT_MS_REG", "BIN_ID"]]
resampled_logs_twt = resampled_logs_twt[cols]

print(f"Paso regular exacto usado: {dt_regular_ms:.6f} ms")
print(f"Muestras originales: {len(resample_df):,}")
print(f"Muestras remuestreadas (TWT regular): {len(resampled_logs_twt):,}")
print(f"Columnas numericas remuestreadas: {len(numeric_cols)}")

resampled_logs_twt[["TWT_MS_REG", "DEPT", "DT", "GR", "RHOB", "TIME_SAMPLING_SEC", "TWT_SAMPLING_MS"]].head()
```

```python
# Plot estilo original pero con eje vertical en tiempo (TWT)
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(9, 10), sharey=True)

plot_twt = resampled_logs_twt.copy()

# Track 1: GAMMA (GR)
axes[0].plot(plot_twt["GR"], plot_twt["TWT_MS_REG"], color="green", linewidth=1.0)
axes[0].set_xlim(0, 200)
axes[0].set_xlabel("GAMMA", fontsize=14, fontweight="bold")
axes[0].set_ylabel("TWT (ms)", fontsize=12)
axes[0].grid(True, alpha=0.3)

# Track 2: DENSITY (RHOB)
axes[1].plot(plot_twt["RHOB"], plot_twt["TWT_MS_REG"], color="red", linewidth=1.0)
axes[1].set_xlim(1.95, 2.95)
axes[1].set_xlabel("DENSITY", fontsize=14, fontweight="bold")
axes[1].grid(True, alpha=0.3)

# Track 3: SONIC (DT)
axes[2].plot(plot_twt["DT"], plot_twt["TWT_MS_REG"], color="darkblue", linewidth=1.0)
axes[2].set_xlim(140, 15)
axes[2].set_xlabel("SONIC", fontsize=14, fontweight="bold")
axes[2].grid(True, alpha=0.3)

# Tiempo hacia abajo para imitar estilo de log
axes[0].invert_yaxis()

for ax in axes:
    ax.tick_params(axis="both", labelsize=10)

plt.tight_layout()
plt.show()
```

`resampled_logs_twt` contiene la version remuestreada en TWT regular usando como paso exacto la media de `TWT_SAMPLING_MS` (sin redondeo).

## Impedancia Acustica (IA) y Coeficiente de Reflexion (RC)

Calculo sobre el segundo set remuestreado en tiempo (`resampled_logs_twt`) usando:

- `IA = VEL_FROM_SONIC_MPS * RHOB`
- `RC = (IA(i+1)-IA(i)) / (IA(i+1)+IA(i))`

```python
# Calculo de IA y RC en el dominio TWT remuestreado
resampled_logs_twt["IA"] = resampled_logs_twt["VEL_FROM_SONIC_MPS"] * resampled_logs_twt["RHOB"]

ia_next = resampled_logs_twt["IA"].shift(-1)
ia_curr = resampled_logs_twt["IA"]
denom = ia_next + ia_curr

# Evitar division por cero y propagar NaN cuando falten datos
resampled_logs_twt["RC"] = np.where(
    np.isfinite(ia_next) & np.isfinite(ia_curr) & np.isfinite(denom) & (denom != 0),
    (ia_next - ia_curr) / denom,
    np.nan,
)

ia_valid = np.isfinite(resampled_logs_twt["IA"]).sum()
rc_valid = np.isfinite(resampled_logs_twt["RC"]).sum()

print(f"Valores validos IA: {ia_valid:,}")
print(f"Valores validos RC: {rc_valid:,}")
print(f"Ultimo RC (esperado NaN): {resampled_logs_twt['RC'].iloc[-1]}")

resampled_logs_twt[["TWT_MS_REG", "RHOB", "VEL_FROM_SONIC_MPS", "IA", "RC"]].head(12)
```

```python
# Visualizacion en TWT para IA y RC
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(7, 10), sharey=True)
plot_ai_rc = resampled_logs_twt.copy()

# Track IA
axes[0].plot(plot_ai_rc["IA"], plot_ai_rc["TWT_MS_REG"], color="purple", linewidth=1.0)
axes[0].set_xlabel("IA", fontsize=12, fontweight="bold")
axes[0].set_ylabel("TWT (ms)", fontsize=12)
axes[0].grid(True, alpha=0.3)

# Track RC
axes[1].plot(plot_ai_rc["RC"], plot_ai_rc["TWT_MS_REG"], color="black", linewidth=1.0)
axes[1].set_xlim(-1, 1)
axes[1].set_xlabel("RC", fontsize=12, fontweight="bold")
axes[1].grid(True, alpha=0.3)

# Tiempo hacia abajo para mantener estilo log
axes[0].invert_yaxis()

for ax in axes:
    ax.tick_params(axis="both", labelsize=10)

plt.tight_layout()
plt.show()
```

## Ondicula Ricker y sismograma sintetico

En esta seccion se genera una ondicula Ricker de 20 Hz con longitud total de 128 ms y muestreo de 2 ms. Luego se realiza la convolucion con la serie `RC` en su muestreo original (sin remuestrear), y se muestra el sintetico.

```python
# Parametros de ondicula
f0_hz = 20.0
wavelet_length_ms = 128.0
dt_conv_ms = 2.0

# Eje temporal de la ondicula (centrado en 0)
half_len_ms = wavelet_length_ms / 2.0
tw_ms = np.arange(-half_len_ms, half_len_ms + dt_conv_ms, dt_conv_ms)
tw_s = tw_ms / 1000.0

# Ondicula de Ricker
pi2f2t2 = (np.pi**2) * (f0_hz**2) * (tw_s**2)
ricker = (1.0 - 2.0 * pi2f2t2) * np.exp(-pi2f2t2)

# Normalizar amplitud maxima a 1
max_abs = np.max(np.abs(ricker))
if max_abs > 0:
    ricker = ricker / max_abs

print(f"Muestras ondicula: {len(ricker)}")
print(f"Duracion ondicula: {wavelet_length_ms:.1f} ms, dt: {dt_conv_ms:.1f} ms")

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(tw_ms, ricker, color="firebrick", linewidth=1.5)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xlabel("Tiempo (ms)")
ax.set_ylabel("Amplitud")
ax.set_title("Ondicula Ricker (20 Hz)")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Serie RC en muestreo original (sin remuestreo)
rc_work = resampled_logs_twt[["TWT_MS_REG", "RC"]].copy()
rc_work = rc_work[np.isfinite(rc_work["TWT_MS_REG"])].sort_values("TWT_MS_REG")

rc_values = rc_work["RC"].astype(float).to_numpy()
twt_values = rc_work["TWT_MS_REG"].astype(float).to_numpy()

# Evitar propagacion de NaN en la convolucion
rc_for_conv = np.nan_to_num(rc_values, nan=0.0, posinf=0.0, neginf=0.0)

# Convolucion discreta conservando muestreo original de RC
synthetic = np.convolve(rc_for_conv, ricker, mode="same")

# Guardar resultado en DataFrame para inspeccion
synthetic_df = pd.DataFrame({
    "TWT_MS": twt_values,
    "RC_ORIGINAL": rc_values,
    "SYNTHETIC": synthetic,
})

print(f"Muestras RC en muestreo original: {len(synthetic_df):,}")
synthetic_df.head()

# Visualizacion RC y sintetico
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(8, 10), sharey=True)

axes[0].plot(synthetic_df["RC_ORIGINAL"], synthetic_df["TWT_MS"], color="black", linewidth=1.0)
axes[0].set_xlabel("RC")
axes[0].set_ylabel("TWT (ms)")
axes[0].set_xlim(-1, 1)
axes[0].set_ylim(1200, 1550)
axes[0].grid(True, alpha=0.3)

axes[1].plot(synthetic_df["SYNTHETIC"], synthetic_df["TWT_MS"], color="navy", linewidth=0.5)
axes[1].set_xlabel("SINTETICO")
axes[1].grid(True, alpha=0.3)

axes[0].invert_yaxis()
plt.tight_layout()
plt.show()
```
